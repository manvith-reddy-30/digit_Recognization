```python
train_df = pd.read_csv('train.csv')

labels = train_df['label']
train_X_df = train_df.drop('label', axis=1)
print(train_df.shape)
train_Y_df = train_df['label']
print(train_Y_df.shape)
test_X_df = pd.read_csv('test.csv')
print(test_X_df.shape)
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train/255
X_test/255
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
model = Sequential()

model.add(Flatten(input_shape=(28,28)))
model.add(Dense(264,activation='relu'))
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(X_train,y_train,batch_size=32,epochs=25,validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
